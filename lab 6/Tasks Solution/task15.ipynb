```python
import pandas as pd
students={"Name":["student1","student2","student3","student4","student5"],
      "Math":[34,54,56,75,33],
      "Science":[24,74,56,72,22],
      "English":[64,23,76,90,31],
    }
df=pd.DataFrame(students)
# Add Total = Math + Science + English.
df["Total"]=df["Math"]+df["Science"]+ df["English"]

# Add Average = Total / 3.
df['Average']=df["Total"]/3

# Add Result = 'Pass' if Average ≥ 50 else 'Fail'.
df['result']=df['Average'].apply(lambda x:"Pass" if x>=50 else "Fail")

# Sort the table by Total in descending order.
df=df.sort_values("Total",ascending=False)

# Print the top 3 students.
print(df.head(3))

# Print the count of Pass vs Fail using .value_counts().
print(df["result"].value_counts())
```
