```python
import matplotlib.pyplot as plt
days=[1,2,3,4,5]
temp=[22,32,43,41,54]
fig,ax=plt.subplots(1,2)
ax[0].bar(days,temp,color="purple")
ax[0].set_title("Temperature Bar Chart")
ax[0].set_xlabel("Days")
ax[0].set_ylabel("Temperature")
ax[0].grid(axis="y")


ax[1].plot(days,temp,marker="o",color="black")
ax[1].set_title("Temperature Line Plot")
ax[1].set_xlabel("Days")
ax[1].set_ylabel("Temperature")
ax[1].grid(True)

plt.tight_layout()
plt.show()
```
